```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Embedding,SimpleRNN
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
text=[
    'I am learning deep learning',
    'i like deep learning',
    'deep learning is fun to learn',
    'i also like machine learning'
]
```

```python
tokenizer=Tokenizer()
tokenizer.fit_on_texts(text)
```

```python
sequences=tokenizer.texts_to_sequences(text)
print(sequences)
```

```python
## Creating input and output later
 ## x=i am learning deep ,_learning =y
X=[]
y=[]
for seq in sequences:
   for i in range(1,len(seq)):
     X.append(seq[:i])
     y.append(seq[i])
```

```python
X=pad_sequences(X)
y = np.array(y, dtype='int32')
y = y.reshape(-1)
```

```python
#vocab_size= unique words
vocab_size=len(tokenizer.word_index)+1
vocab_size
```

```python
##Build rnn model
model=Sequential([
    Embedding(vocab_size, 10, input_length=X.shape[1]),SimpleRNN(32),Dense(vocab_size,activation='softmax')
])
```

```python
model.compile(loss='sparse_categorical_crossentropy', optimizer='adam')
```

```python
model.fit(X,y,epochs=100)
```

```python
inputtext='I am'
seq = tokenizer.texts_to_sequences([inputtext])[0]

# 2. Pad sequence (same length as training)
seq = pad_sequences([seq], maxlen=X.shape[1])

# 3. Predict probabilities
pred = model.predict(seq)


```

```python
pred_index=np.argmax(pred)
```

```python
for word,index in tokenizer.word_index.items():
  if index== pred_index:
    print('Next word: ',word)
```
